# Data Cleaning & Preprocesing

```python
import sys
import os

sys.path.append(os.path.abspath(".."))

import pandas as pd

from src.data.load_data import load_data
from src.data.preprocess import encode_binary, encode_categorical

df = load_data('../data/raw/telco_customer_churn.csv')

df.info()
```

This dataset has 7043 entries (customers) and 21 columns, where 17 are categorical variables and 4 are numeric variables.
Looking into the data type from each column, it is possible to see that TotalCharges is parsed as object despite being a numeric field. This is the primary dtype issue that needs to be flagged and documented here. The full dtype map is: 1 float64, 2 int64, 18 object.

## Fix Data Types

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.info()
```

## Handle Missing Values

```python
df.isnull().sum()
```

TotalChages has 11 missing values, therefore it's important to establish a method to deal with this. In this case to handle the missing data, it will be used the median of TotalCharges.

```python
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
df.isnull().sum()
```

## Hnadling Duplicate Values

```python
df.duplicated().sum()
```

In this dataset, there are no duplicates.

## Clean Categorical Values
In order to see if thera are inconsistent labels, extra spaces or weird categories, it is important to look to the unique values of each categorical column.

```python
for col in df.select_dtypes(include="object"):
    print(col, df[col].unique())
```

## Drop Irrelevant Fatures

Looking into all the variables, it is possible to see that features CustomerID may be irrelevant to the study.

```python
df = df.drop(columns=["customerID"])
```

## Encode Variables
One of the most important steps in this model is the encoding of the categorical variables.

### Convert Target Variable (Churn)

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
```

### Convert Binary Columns
Columns like:
- Partner
- Dependents
- PhoneService
- PaperlessBilling

```python
df, binary_columns = encode_binary(df)

df.head()
```

### Convert Categorical Variables
Some variables have more than two categories, such as MultipleLines, InternetService, OnlineSecurity, and so on. Therefore, it is necessary to apply a type of encoding called one-hot encoding to these variables, making them easier to use in machine learning models.

```python
df, categorical_columns = encode_categorical(df)

df.head()
```

This one-hot encoding transforms the dataset and every categorical variable is now a binary variable with True/False. To have consistency, compability and readability, this features are going to be convert in 1/0.

```python
bool_cols = df.select_dtypes(include="bool").columns

df[bool_cols] = df[bool_cols].astype(int)

df.head()
```

## Save Clean Dataset

```python
df.to_csv("../data/processed/telco_clean.csv", index=False)
```
